This notebook defines a 2D velocity vector and visualizes its vector field using numpy and matplotlib. It calculates and plots the direction and magnitude of the vector field over a specified grid.

2D, Velocity vector V = (u,v) = (0.5 + 0.8x) i + (1.5 - 0.8 y) j

```python
import numpy as np
import matplotlib.pyplot as plt

def u(x):
    return 0.5 + 0.8*x

def v(y):
    return 1.5 + 0.8*y

def magnitude(U, V):
    return np.sqrt(U**2 + V**2)

# grid
grid_x = np.linspace(-3, 3, 30)
grid_y = np.linspace(-5, 5, 30)

X, Y = np.meshgrid(grid_x, grid_y)

# field values
U = u(X)
V = v(Y)

# direction vectors (normalized)
Mag = magnitude(U, V)
U_dir = U / Mag
V_dir = V / Mag

plt.figure(figsize=(10, 7))
plt.quiver(X, Y, U_dir, V_dir, Mag, cmap="viridis")  # color by magnitude
plt.colorbar(label="Magnitude")

plt.title("Vector field")
plt.xlabel("x")
plt.ylabel("y")
plt.grid(True)
plt.axis("equal")
plt.show()
```
